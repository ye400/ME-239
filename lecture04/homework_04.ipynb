## Problem 2 - Visual analysis of an airfoil experiment

In this problem, you are going to repeat what you did in Problem 1, but without my guidance!

The dataset we are going to use is the [Airfoil Self-Noise Data Set](https://archive.ics.uci.edu/dataset/291/airfoil+self+noise)
From this reference, the descreption of the dataset is as follows:

> The NASA data set comprises different size NACA 0012 airfoils at various wind tunnel speeds and angles of attack. The span of the airfoil and the observer position were the same in all of the experiments.
>
> Attribute Information:
> This problem has the following inputs:
> 1. Frequency, in Hertzs.
> 2. Angle of attack, in degrees.
> 3. Chord length, in meters.
> 4. Free-stream velocity, in meters per second.
> 5. Suction side displacement thickness, in meters.

> The only output is:
> 6. Scaled sound pressure level, in decibels.

Before we start, let's download and load the data.
I am going to put them in a dataframe for you.

````python
!curl -O 'https://archive.ics.uci.edu/ml/machine-learning-databases/00291/airfoil_self_noise.dat'
import numpy as np
import pandas as pd
raw_data = np.loadtxt('airfoil_self_noise.dat')
df = pd.DataFrame(raw_data,
                  columns=['Frequency', 'Angle_of_attack', 'Chord_length',
                            'Velocity', 'Suction_thickness', 'Sound_pressure'])
df
````

+ Do the histogtrams of all variables. Use as many code segments you need below to plot the histogram of each variable in a different plot. Make sure you label the axes correctly.

````python
labels = {
    'Frequency': 'Frequency (Hz)',
    'Angle_of_attack': 'Angle of attack (degrees)',
    'Chord_length': 'Chord length (m)',
    'Velocity': 'Free-stream velocity (m/s)',
    'Suction_thickness': 'Suction-side displacement thickness (m)',
    'Sound_pressure': 'Scaled sound pressure level (dB)'
}

for column in df.columns:
    plt.figure()
    plt.hist(df[column], bins=20, edgecolor='black')
    plt.xlabel(labels[column])
    plt.ylabel('Count')
    plt.title(labels[column])
    plt.show()
````

+ Do the scatter plot between all input variables. This will give you an idea of the range of experimental conditions. Are there any holes in the experimental dataset, i.e., places where you have no data?

````python
inputs = [
    'Frequency',
    'Angle_of_attack',
    'Chord_length',
    'Velocity',
    'Suction_thickness'
]

for x, y in [(x, y) for x in inputs for y in inputs if x != y]:
    plt.figure()
    plt.scatter(df[x], df[y])
    plt.xlabel(labels[x])
    plt.ylabel(labels[y])
````

Yes, there are gaps because the experiments used selected input values rather than every possible combination.

+  Do the scatter plot between each input variable and the output. This will give you an idea of the relationship between each input and the output. Do you observe any obvious patterns?

````python
for x in inputs:
    plt.figure()
    plt.scatter(df[x], df['Sound_pressure'])
    plt.xlabel(labels[x])
    plt.ylabel(labels['Sound_pressure'])
````

Sound pressure usually decrease as frequency and suction-side displacement thickness increase, but there is considerable scatter. Angle of attack and chord length show weaker and complicated patterns. Velocity has a weak positive relationship with sound pressure overall, but the sound levels overlap a decent amount between the tested velocities. These plots suggest that no single input explains all the variation in sound pressure.

+ Now pick the two input variables you think are the most important and do the scatter plot between them using the output to color the points (see the last question of Problem 1). Feel free to repeat it with more than two pairs of inputs if you want. Briefly discuss your findings.

````python
points = plt.scatter(
    df['Frequency'],
    df['Suction_thickness'],
    c=df['Sound_pressure'],
    cmap='viridis',
    s=25,
    alpha=0.8
)

plt.colorbar(points, label=labels['Sound_pressure'])
plt.xlabel(labels['Frequency'])
plt.ylabel(labels['Suction_thickness'])
plt.show()
````

I selected frequency and suction-side displacement thickness because both show noticeable relationships with sound pressure. The colored plot shows that lower sound levels tend to occur at higher frequencies and greater displacement thicknesses. However, the pattern is not uniform, suggesting that the other experimental variables also influence sound pressure.
